Handling Categorical Missing Values

These types of columns can be imputed by 2 techniques:
1. Filling with most frequent values
2. Fill the NaN values with 'Missing'

To fill up the NaN values with most frequent values we use mode().


For example, in a column called Country there are UAE, Japan, Germany categories. There are some NaN values and with the frequency of UAE is 30, Japan is 13 and Germany is 18 so the values will be replaced with UAE.


One thing should be taken care of is the difference in the highet and second highest frequency category should be high.


How should you decide?

Consider these three things:
1. Frequency gap: Is the most frequent category clearly ahead of the second most frequent category?
2. Missing percentage: How many values are missing? If 5 values are missing, mode imputation may have little impact. If 300 are missing, it may significantly distort the distribution.
3. Reason for missingness: Are values missing randomly, or could the missingness be related to the actual category?
4. Mode should be used when the missing value percentage is 5-10% of the overall data only.

Next technique is Filling NaN with 'Missng'

If the NaN values percentage is high or the NaN values are following a particular pattern (no randomness) then we fill NaN values with 'Missing' creating a new category
